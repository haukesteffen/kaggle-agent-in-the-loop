# **SETUP**

```python
import pathlib
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

PROJECT_ROOT = pathlib.Path().absolute().parent

train = pd.read_parquet(PROJECT_ROOT / "data" / "train.parquet")
test = pd.read_parquet(PROJECT_ROOT / "data" / "test.parquet")
cv = pd.read_parquet(PROJECT_ROOT / "data" / "cv.parquet")

ohe = OneHotEncoder(sparse_output=False, handle_unknown="ignore").set_output(transform="pandas")
```

# **ONE-HOT-ENCODING HIGH-CARDINALITY DRIVER**

```python
train_subset = train[["Driver"]]
test_subset = test[["Driver"]]

train_out = ohe.fit_transform(train_subset).astype("int8")
test_out = ohe.transform(test_subset).astype("int8")
```

# **EXPORT**

```python
feature_name = "006-one-hot-driver-high-cardinality"
(PROJECT_ROOT / "data" / "features" / feature_name).mkdir(exist_ok=True)

train_out.to_parquet(PROJECT_ROOT / "data" / "features" / feature_name / "train.parquet")
test_out.to_parquet(PROJECT_ROOT / "data" / "features" / feature_name / "test.parquet")
```
