# **SETUP**

```python
import pathlib
import pandas as pd

PROJECT_ROOT = pathlib.Path().absolute().parent

train = pd.read_csv(PROJECT_ROOT / "data" / "raw" / "train.csv", index_col=0)
test = pd.read_csv(PROJECT_ROOT / "data" / "raw" / "test.csv", index_col=0)
original = pd.read_csv(PROJECT_ROOT / "data" / "raw" / "original.csv", index_col=False)
```

# **FIX DTYPES**

```python
types = {
    "Driver":                   "category",
    "Compound":                 "category",
    "Race":                     "category",
    "Year":                     "category",
    "PitStop":                  "int8",
    "LapNumber":                "int8",
    "Stint":                    "int8",
    "TyreLife":                 "int8",
    "Position":                 "int8",
    "LapTime (s)":              "float64",
    "LapTime_Delta":            "float64",
    "Cumulative_Degradation":   "float64",
    "RaceProgress":             "float64",
    "Position_Change":          "int8",
}
test = test.astype({"Year":"string"}).astype(types)
types["PitNextLap"] = "int8"
train = train.astype({"Year":"string"}).astype(types)
original = original.astype({"Year":"string"}).astype(types)
```

# **EXPORT**

```python
train.to_parquet(PROJECT_ROOT / "data" / "train.parquet", index=True)
test.to_parquet(PROJECT_ROOT / "data" / "test.parquet", index=True)
original.to_parquet(PROJECT_ROOT / "data" / "original.parquet", index=True)
```
